## 4. Missing Values & Whitespace Detection

```python
# Standard null checks
null_counts = df.isnull().sum()
null_percentages = (null_counts / len(df)) * 100
missing_df = pd.DataFrame({"Null Count": null_counts, "Null Percentage (%)": null_percentages.round(2)})
print(missing_df)

# Inspect empty strings / whitespace values across object columns
space_counts = {}
for col in df.select_dtypes(include=["object"]).columns:
    cnt = df[col].astype(str).str.strip().eq("").sum()
    if cnt > 0:
        space_counts[col] = cnt
print(f"Columns with empty/whitespace strings: {space_counts}")
```

## 5. Duplicate Records Check

```python
dup_count = df.duplicated().sum()
print(f"Exact Duplicate Rows: {dup_count}")
```

## 6. Categorical Columns & Unique Values

```python
cat_cols = df.select_dtypes(include=["object"]).columns.drop(["customerID", "Churn"], errors="ignore")
for col in cat_cols:
    print(f"=== {col} ===")
    print(df[col].value_counts(dropna=False))
    print()
```

## 7. Target Variable Inspection (`Churn`)

```python
target_counts = df["Churn"].value_counts(dropna=False)
target_pcts = (df["Churn"].value_counts(normalize=True, dropna=False) * 100).round(2)
target_df = pd.DataFrame({"Count": target_counts, "Percentage (%)": target_pcts})
print(target_df)
```

## 8. Data Quality & Type Discrepancies Investigation

```python
# Inspect the 11 records where TotalCharges is whitespace
blank_tc_df = df[df["TotalCharges"].str.strip() == ""]
print(f"Number of records with blank TotalCharges: {len(blank_tc_df)}")
print(blank_tc_df[["customerID", "tenure", "MonthlyCharges", "TotalCharges", "Churn"]])
```

## 9. Key Findings & Observations

1. **Dataset Dimensions:** 7,043 rows and 21 columns.
2. **Target Variable (`Churn`):** Imbalanced class distribution — 5,174 `No` (73.46%) vs 1,869 `Yes` (26.54%).
3. **Identifier Column:** `customerID` (7,043 unique string identifiers).
4. **Data Type Discrepancy:** `TotalCharges` is stored as `object` (string) due to 11 empty whitespace values (`" "`). These 11 records all have `tenure == 0` (newly onboarded customers).
5. **Missing / Duplicate Status:** 0 standard NaN values; 0 duplicate rows; 11 implicit missing values in `TotalCharges`.
6. **Categorical Features:** Multi-level indicators (e.g. `No internet service` or `No phone service`) present across service sub-columns.
