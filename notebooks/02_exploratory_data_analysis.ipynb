### Objective
Explore the dataset to uncover empirical distributions, churn associations, feature relationships, and potential outliers. All analyses are strictly descriptive without training models, applying transformations, or claiming causality.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Styling settings
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["font.sans-serif"] = "DejaVu Sans"

# Ensure figures directory exists
os.makedirs("../reports/figures", exist_ok=True)

# Load raw dataset
DATA_PATH = "../data/raw/WA_Fn-UseC_-Telco-Customer-Churn.csv"
df = pd.read_csv(DATA_PATH)
print(f"Raw dataset successfully loaded. Shape: {df.shape}")
```

## 1. Dataset Overview

This section outlines the basic structural dimensions based on Phase 2 findings.
* **Total Observations:** 7,043
* **Total Attributes:** 21
* **Target Column:** `Churn`
* **Identifier Column:** `customerID` (7,043 unique alphanumeric strings)
* **Numerical Features (3):** `tenure`, `MonthlyCharges`, `TotalCharges` (coerced numeric for analysis)
* **Categorical Features (16):** `gender`, `SeniorCitizen`, `Partner`, `Dependents`, `PhoneService`, `MultipleLines`, `InternetService`, `OnlineSecurity`, `OnlineBackup`, `DeviceProtection`, `TechSupport`, `StreamingTV`, `StreamingMovies`, `Contract`, `PaperlessBilling`, `PaymentMethod`

```python
# Create temporary dataframe for EDA analysis with TotalCharges converted to numeric
df_eda = df.copy()
df_eda["TotalCharges_num"] = pd.to_numeric(df_eda["TotalCharges"], errors="coerce")
print(f"Dataset shape for EDA analysis: {df_eda.shape}")
```

## 2. Target Variable Analysis (`Churn`)

We inspect the distribution of the target variable `Churn` using raw text categories (`"No"` / `"Yes"`).

```python
target_counts = df["Churn"].value_counts()
target_pcts = (df["Churn"].value_counts(normalize=True) * 100).round(2)
target_df = pd.DataFrame({"Count": target_counts, "Percentage (%)": target_pcts})
print(target_df)

# Visualization
plt.figure(figsize=(6, 4))
ax = sns.countplot(data=df, x="Churn", hue="Churn", palette=["#2b5c8f", "#d9534f"], legend=False)
plt.title("Target Variable Distribution (Churn)", fontsize=13, pad=10)
plt.xlabel("Churn Status", fontsize=11)
plt.ylabel("Customer Count", fontsize=11)
total = len(df)
for p in ax.patches:
    h = p.get_height()
    pct = h / total * 100
    ax.annotate(f"{int(h):,}\n({pct:.1f}%)", (p.get_x() + p.get_width()/2., h/2),
                ha="center", va="center", color="white", fontweight="bold", fontsize=11)
plt.tight_layout()
plt.savefig("../reports/figures/target_distribution.png", dpi=150)
plt.show()
```

### Target Analysis Interpretation
* **Retained (`No`):** 5,174 customers (73.46%)
* **Churned (`Yes`):** 1,869 customers (26.54%)
* **Class Imbalance:** The target variable is imbalanced with a ratio of approximately 2.77 retained customers for every 1 churned customer.

## 3. Numerical Features Analysis

We analyze the distributions, central tendencies, and spreads of `tenure`, `MonthlyCharges`, and `TotalCharges`.

```python
# Descriptive statistics for numerical features
num_cols = ["tenure", "MonthlyCharges", "TotalCharges_num"]
print(df_eda[num_cols].describe().round(2))

# Histograms and KDE plots
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
sns.histplot(df_eda["tenure"], kde=True, ax=axes[0], color="#2b5c8f", bins=30)
axes[0].set_title("Tenure Distribution (Months)", fontsize=12)
axes[0].set_xlabel("Tenure (Months)")
axes[0].set_ylabel("Customer Count")

sns.histplot(df_eda["MonthlyCharges"], kde=True, ax=axes[1], color="#4682b4", bins=30)
axes[1].set_title("Monthly Charges Distribution ($)", fontsize=12)
axes[1].set_xlabel("Monthly Charges ($)")
axes[1].set_ylabel("Customer Count")

sns.histplot(df_eda["TotalCharges_num"].dropna(), kde=True, ax=axes[2], color="#5f9ea0", bins=30)
axes[2].set_title("Total Charges Distribution ($)", fontsize=12)
axes[2].set_xlabel("Total Charges ($)")
axes[2].set_ylabel("Customer Count")

plt.tight_layout()
plt.savefig("../reports/figures/numerical_distributions.png", dpi=150)
plt.show()
```

### Numerical Distributions Interpretation
* **`tenure`:** Bimodal distribution with peaks near 1 month (new customers) and 72 months (long-term loyal customers). Mean tenure is 32.37 months (median 29.0).
* **`MonthlyCharges`:** Range from $18.25 to $118.75 with a mean of $64.76 (median $70.35). High concentration near $20 representing basic phone-only plans.
* **`TotalCharges`:** Positively skewed distribution ranging from $18.80 to $8,684.80 with a mean of $2,555.34 (median $1,683.60).

## 4. Categorical Features Analysis

We inspect key categorical features and their frequency counts.

```python
cat_summary_cols = ["Contract", "InternetService", "PaymentMethod", "TechSupport", "SeniorCitizen", "PaperlessBilling"]
for col in cat_summary_cols:
    print(f"=== {col} Distribution ===")
    vc = df[col].value_counts()
    pct = (df[col].value_counts(normalize=True) * 100).round(2)
    print(pd.DataFrame({"Count": vc, "Percentage (%)": pct}))
    print()
```

### Churn vs Numerical Features Interpretation
* **Tenure:** Churned customers exhibit a substantially lower median tenure (**10.0 months**) compared to retained customers (**38.0 months**).
* **Monthly Charges:** Churned customers exhibit higher median monthly charges (**$79.65**) compared to retained customers (**$64.43**).
* **Total Charges:** Churned customers have a lower median total charges (**$703.55**) than retained customers (**$1,683.60**), driven primarily by shorter overall tenure.

## 8. Outlier Investigation

We evaluate numerical features using the Interquartile Range (IQR) method.

```python
for col in ["tenure", "MonthlyCharges", "TotalCharges_num"]:
    series = df_eda[col].dropna()
    q25, q75 = series.quantile(0.25), series.quantile(0.75)
    iqr = q75 - q25
    lower_bound = q25 - 1.5 * iqr
    upper_bound = q75 + 1.5 * iqr
    outliers = series[(series < lower_bound) | (series > upper_bound)]
    print(f"{col} — IQR: {iqr:.2f} | Bounds: [{lower_bound:.2f}, {upper_bound:.2f}] | Outliers Count: {len(outliers)}")
```

### Outlier Findings
* **Zero IQR Outliers:** None of `tenure`, `MonthlyCharges`, or `TotalCharges` contain statistical outliers under the 1.5 × IQR rule.
* **Data Integrity:** All numerical values fall within realistic domain ranges (tenure 0–72 months, monthly charges $18.25–$118.75). No automatic deletion is required.

## 9. Class Imbalance Analysis

* **Target Distribution:** 73.46% Retained (`No`) vs. 26.54% Churned (`Yes`).
* **Evaluation Impact:** A naive baseline model predicting `No` for all customers would achieve 73.46% accuracy. Therefore, **accuracy alone is an unreliable evaluation metric**.
* **Model Metrics Required:** Model selection during evaluation must prioritize Precision, Recall, F1-Score, and ROC-AUC.

## 10. Key Empirical Observations & Phase 4 Implications

### Observed Empirical Patterns
1. **Contract Type:** Customers on Month-to-month contracts exhibited a **42.71%** churn rate, compared to **11.27%** for One-year and **2.83%** for Two-year contracts.
2. **Internet Service:** Fiber optic subscribers showed a **41.89%** churn rate compared to **18.96%** for DSL subscribers.
3. **Tenure Duration:** Customers who churned had a median tenure of **10.0 months**, compared to **38.0 months** for retained customers.
4. **Payment Method:** Customers using Electronic Check exhibited a **45.29%** churn rate, whereas automatic payment methods averaged ~16%.
5. **Tech Support:** Customers without Tech Support experienced a **41.64%** churn rate compared to **15.17%** for those with Tech Support.

### Implications for Phase 4 (Preprocessing)
* **Type Conversion & Imputation:** `TotalCharges` must be coerced to `float64` and the 11 missing values imputed (or zeroed based on `tenure == 0`).
* **Categorical Encoding:** Multi-category variables (`Contract`, `PaymentMethod`, service options) require encoding strategies (e.g. One-Hot Encoding).
* **Identifier Handling:** `customerID` must be excluded from feature vectors.
* **Target Encoding:** `Churn` should be mapped to `0` (No) and `1` (Yes) in Phase 4 preprocessing pipelines.
